<h1>Rule for Selection of stock</>
<h4> Buy the stock that's price is below the 2sd of its average price in last 60 days. </>
<h4> Sell the stock that's price is above the 1.5sd of its average price in last 60 days. </>
<h4> Buy and sell has at least 4 day's gap <h2>

```python
# portfolio
portfolio = []

# add stock in the portfolio
for s in trading_codes['symbol']:
    
    df = db[db['symbol']==s]

    # remove the values with 0
    df = df[df['ltp']!=0]
    date = "2023-01-01"

    # get length upto the date




    if date in df.date.to_list():
        row = df[df["date"]==date].index.tolist()[0]

        today_price = df[df["date"]==date]['ltp'].values[0]

        # before 3 dyas for 60 days calc sd and average
        last_days_ltp = df['ltp'].loc[row+3: row + 62]

        last_days_sd = last_days_ltp.std()
        last_days_avg = last_days_ltp.mean()


        # check if today's price is below 2sd of last 60 days average

        if today_price <= last_days_avg - 20*last_days_sd:
            print(f"Buy the stock {s} at Taka {today_price} on {date}")

        
        # row = row - 1
        # date = df.loc[row]['date']

    # else: print(f"date not found for {s}")
    
# calculate past 60 day sd

```
